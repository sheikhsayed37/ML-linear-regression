## Question 3 (20 Marks)

Implement **Simple Linear Regression** using **only NumPy** (no Scikit-Learn allowed):
- Compute slope (`m`) and intercept (`c`) using the Batch Gradient Descent
- Predict values for the test set
- Print the learned `m` and `c` values

Use `Square_Footage` as feature (X) and `House_Price` as target (y).

````python
#Question 3
x=df["Square_Footage"].values
y=df["House_Price"].values
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42)


def compute_cost(x,y,W,b):
    m=x.shape[0]
    cost=0
    for i in range(m):
        pred_y=W*x[i]+b

        error =  pred_y - y[i]

        cost = cost + error**2

    return cost / (2 * m)

def gradient(x,y,W,b):
  m=x.shape[0]
  dj_dw=0
  dj_db=0

  for i in range(m):
    pred_y= W*x[i]+b

    error = pred_y - y[i]
    dj_dw = dj_dw +(error*x[i])
    dj_db = dj_db + error

  return dj_dw / m, dj_db / m

def gradient_descent(x,y,w_input,b_input,alpha,num_iters):
    W = w_input
    b = b_input

    cost_memo = []
    iteration = []


    for i in range(num_iters):

      dj_dw,dj_db=gradient(x,y,W,b)

      W=W-alpha*dj_dw

      b=b-alpha*dj_db

      cost=compute_cost(x,y,W,b)
      cost_memo.append(cost)
      iteration.append(i)
      if i % 100 == 0:
        print(f"Iteration {i}: Cost = {cost:.4f}, w = {W:.4f}, b = {b:.4f}")




    return W,b,cost_memo,iteration

w_final,b_final,cost_memo,iteration=gradient_descent(x_train,y_train,w_input=0,
    b_input=0,
    num_iters=1000,
    alpha=0.0000000001
)


for i in range(len(x_test)):

  pred_y=w_final*x_test[i]+b_final
   # x _test final weight gola use kore prediction
  print("Actual:", y_test[i], " Predicted:", pred_y)

print("\nSlope (m) =", w_final)
print("Intercept (c) =", b_final)


````

Build a **ColumnTransformer** that applies:

## Question 5 (20 Marks)

Build a complete **Pipeline** using Scikit-Learn that includes:
- The `ColumnTransformer`
- `SGDRegressor` as the final estimator
- Train the pipeline and evaluate using RMSE and R² score
- Print predicted vs actual values for the first 10 test samples

````python
from sklearn.linear_model import SGDRegressor
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, r2_score

x=df.drop("House_Price",axis=1)
y=df["House_Price"]

x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42)

ETA = 0.001
MAX_ITER =10000
ALPHA = 0.0001
sgd_pipeline=Pipeline(
    steps=[
        ('prepocessor',prepocessor),
        ('Modal2',SGDRegressor(eta0=ETA,max_iter=MAX_ITER,alpha=ALPHA))

    ]

)

sgd_pipeline.fit(x_train,y_train)
train_pred=sgd_pipeline.predict(x_train)
test_pred = sgd_pipeline.predict(x_test)

print("Train R2 Score:", r2_score(y_train, train_pred))
print("Test R2 Score:", r2_score(y_test, test_pred))
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, test_pred)),4))

````

## 1. ETA = 0.001

**ETA = Learning Rate**

এটা বলে:

> "একবারে কতটুকু শিখবি?"

ধরো তুমি সিঁড়ি দিয়ে নামছো।

* ETA বেশি → ৫টা ধাপ একসাথে লাফ দিবে 😅
* ETA কম → ১ ধাপ করে নামবে 🙂

```python
ETA = 0.001
```

মানে SGD আস্তে আস্তে weight update করবে।

Gradient Descent-এ:

```text
new_weight = old_weight - ETA × gradient
```

---

## 2. MAX_ITER = 3000

এটা বলে:

> "সর্বোচ্চ কতবার practice করতে পারবি?"

ধরো ছাত্রকে 3000 বার অনুশীলন করতে দেওয়া হয়েছে।

```python
MAX_ITER = 3000
```

মানে SGD সর্বোচ্চ 3000 বার weight update করবে।

Manual GD-তে যেমন:

```python
for i in range(3000):
```

একই ব্যাপার।

---

## 3. ALPHA = 0.0001

এটা বলে:

> "বেশি চালাকি করিস না, weight খুব বড় করিস না।"

ALPHA হলো **regularization**।

যদি model weight খুব বড় করতে শুরু করে:

```text
w1 = 50000
w2 = -80000
```

তাহলে ALPHA penalty দেয়।

ফলে model একটু simple থাকে এবং overfitting কমে।

---

## সবচেয়ে সহজ ভাষায়

| Parameter | কাজ                                     |
| --------- | --------------------------------------- |
| ETA       | কত দ্রুত শিখবে                          |
| MAX_ITER  | কতবার শিখবে                             |
| ALPHA     | overfitting কমাতে weight নিয়ন্ত্রণ করবে |

---

এক লাইনে মনে রাখো:

```text
ETA      → শেখার গতি
MAX_ITER → শেখার সময়/বার
ALPHA    → বেশি মুখস্থ করা আটকায়
```

SGDRegressor-এর performance অনেক সময় এই ৩টা hyperparameter-এর উপর নির্ভর করে।


#ETA (Learning Rate):
ETA model কত দ্রুত শিখবে তা নির্ধারণ করে। ছোট ETA হলে model ধীরে ধীরে weight update করে, আর বড় ETA হলে দ্রুত update করে।

#MAX_ITER:
MAX_ITER নির্ধারণ করে SGDRegressor সর্বোচ্চ কতবার weight update করবে। বেশি iteration হলে model শেখার জন্য বেশি সুযোগ পায়।

#$ALPHA:
ALPHA হলো regularization parameter। এটি weight খুব বড় হয়ে যাওয়া রোধ করে এবং overfitting কমাতে সাহায্য করে।

সংক্ষেপে:

ETA → শেখার গতি
MAX_ITER → কতবার শিখবে
ALPHA → Overfitting নিয়ন্ত্রণ করে এবং model-কে simpler রাখে।
