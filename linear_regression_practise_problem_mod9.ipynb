## Cost Function
1. Cost Function কেন দরকার?
2. Prediction Error কী?
3. Model error কমানো কেন গুরুত্বপূর্ণ?
4. Large error হলে Cost Function এর value কেমন হবে?
5. Cost কমানোর process এর নাম কী?


1. Cost Function কেন দরকার?

Model কতটা ভুল prediction করছে তা মাপার জন্য।

2. Prediction Error কী?

Actual Value এবং Predicted Value এর পার্থক্য।

Error = Actual - Predicted

3. Model error কমানো কেন গুরুত্বপূর্ণ?

কারণ error কম হলে prediction বেশি accurate হয়।

4. Large error হলে Cost Function এর value কেমন হবে?

Large Error → Large Cost

5. Cost কমানোর process এর নাম কী?

Optimization

সাধারণত Gradient Descent ব্যবহার করা হয়।

```python

# example usage
X = np.array([800, 1000, 1200, 1500, 1800])
w = 50
b = 10000

def make_prediction(X, w, b):
    """
    Calculates predictions using linear regression: y = wx + b
    w: weight (slope)
    b: bias (intercept)
    """

    m = X.shape[0]
    pred_list = np.zeros(m)

    for i in range(m):
        pred_list[i] = w * X[i] + b

    return pred_list


y_pred = make_prediction(X, w, b)

print(y_pred)
```
